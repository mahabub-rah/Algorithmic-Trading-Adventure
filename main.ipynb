```python
import yfinance as yf  #access historical market data
import pandas as pd
```

```python
ticker_symbol = 'AAPL'
ticker = yf.Ticker(ticker_symbol)
print(ticker)
```

```python
data = yf.download(
                tickers='AMZN', 
                start="2018-01-01", 
                end="2023-12-31",
                progress= False,
                threads=False
                )
```

```python
data.isnull().sum()
```

```python
data.duplicated().sum()
```

```python
data = data.drop_duplicates(keep='first')
data = data.ffill()
```
